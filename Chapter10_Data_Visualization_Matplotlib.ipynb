## What is Matplotlib?

Matplotlib is a flexible, widely-used plotting library. Colab already has it installed, so
these cells run without any setup.


### Making a line graph and a scatter plot


```python
import matplotlib.pyplot as plt

x_values = [0, 1, 2, 3, 4, 5]
squares = [0, 1, 4, 9, 16, 25]

plt.plot(x_values, squares)
plt.show()
```

```python
x_values = list(range(1000))
squares = [x**2 for x in x_values]

plt.scatter(x_values, squares, s=10)
plt.show()
```

### Customizing plots: titles, labels, and axes


```python
x_values = list(range(1000))
squares = [x**2 for x in x_values]

plt.scatter(x_values, squares, s=10)

plt.title("Square Numbers", fontsize=24)
plt.xlabel("Value", fontsize=18)
plt.ylabel("Square of Value", fontsize=18)
plt.tick_params(axis='both', which='major', labelsize=14)

plt.axis([0, 1100, 0, 1_100_000])
plt.show()
```

### Using a colormap

A colormap varies point colors from one shade to another, based on a value for each point.


```python
plt.scatter(x_values, squares, c=squares, cmap=plt.cm.Blues, edgecolor='none', s=10)
plt.title("Square Numbers", fontsize=24)
plt.show()
```

### Emphasizing points

You can re-plot specific points on top of the main data to make them stand out.


```python
plt.scatter(x_values, squares, c='blue', edgecolor='none', s=20)
plt.scatter(x_values[0], squares[0], c='green', edgecolor='none', s=100)
plt.scatter(x_values[-1], squares[-1], c='red', edgecolor='none', s=100)

plt.title("Square Numbers", fontsize=24)
plt.show()
```

### Removing axes (modern API)

The original cheat sheet used `plt.axes().get_xaxis()`, which is deprecated. The current,
recommended way is to grab the axes object explicitly with `plt.subplots()` or `plt.gca()`.


```python
fig, ax = plt.subplots()
ax.scatter(x_values, squares, s=10)
ax.get_xaxis().set_visible(False)
ax.get_yaxis().set_visible(False)
plt.show()
```

### Setting a custom figure size, and saving a plot


```python
plt.figure(figsize=(10, 6), dpi=128)
plt.scatter(x_values, squares, s=10)
plt.savefig('squares.png', bbox_inches='tight')
plt.show()
print("Saved to squares.png")
```

## Plotting two sets of data, and filling the space between them


```python
x_values = list(range(11))
squares = [x**2 for x in x_values]
cubes = [x**3 for x in x_values]

plt.scatter(x_values, squares, c='blue', edgecolor='none', s=20)
plt.scatter(x_values, cubes, c='red', edgecolor='none', s=20)
plt.fill_between(x_values, squares, cubes, facecolor='blue', alpha=0.25)
plt.axis([0, 11, 0, 1100])
plt.show()
```

## Working with dates and times


```python
from datetime import datetime as dt

today = dt.now()
date_string = today.strftime('%m/%d/%Y')
print(date_string)

new_years = dt.strptime('1/1/2030', '%m/%d/%Y')
print(new_years.strftime('%B %d, %Y'))
```

```python
dates = [
    dt(2024, 6, 21), dt(2024, 6, 22),
    dt(2024, 6, 23), dt(2024, 6, 24),
]
highs = [57, 68, 64, 59]

fig, ax = plt.subplots(figsize=(10, 6), dpi=128)
ax.plot(dates, highs, c='red')
ax.set_title("Daily High Temps", fontsize=24)
ax.set_ylabel("Temp (F)", fontsize=16)
fig.autofmt_xdate()
plt.show()
```

### Multiple plots in one figure, sharing an axis


```python
x_vals = list(range(11))
squares = [x**2 for x in x_vals]
cubes = [x**3 for x in x_vals]

fig, axarr = plt.subplots(2, 1, sharex=True, figsize=(6, 8))
axarr[0].scatter(x_vals, squares)
axarr[0].set_title('Squares')
axarr[1].scatter(x_vals, cubes, c='red')
axarr[1].set_title('Cubes')
plt.show()
```
