## Conditional tests

A conditional test is an expression that evaluates to `True` or `False`. Python uses these
values to decide whether the code inside an `if` statement should run.


```python
car = 'bmw'
print(car == 'bmw')
print(car != 'audi')

car = 'Audi'
print(car.lower() == 'audi')   # ignoring case
```

### Numerical comparisons


```python
age = 19
print(age == 18)
print(age != 18)
print(age < 21)
print(age <= 21)
print(age > 21)
print(age >= 21)
```

### Checking multiple conditions with `and` / `or`


```python
age_0, age_1 = 22, 18
print(age_0 >= 21 and age_1 >= 21)
print(age_0 >= 21 or age_1 >= 21)
```

### Conditional tests with lists


```python
players = ['al', 'bea', 'cyn', 'dale']
print('al' in players)
print('eric' not in players)
```

## if / elif / else

Several kinds of `if` statements exist. Your choice depends on how many conditions you need to
test. You can have as many `elif` blocks as you need, and the `else` block is always optional.


```python
age = 12

if age < 4:
    price = 0
elif age < 18:
    price = 5
else:
    price = 10

print(f"Your cost is ${price}.")
```

## While loops

A `while` loop repeats a block of code as long as a certain condition is `True`.


```python
current_value = 1
while current_value <= 5:
    print(current_value)
    current_value += 1
```

### Letting the user choose when to quit
In Colab, `input()` still works -- it just prompts inline below the cell.


```python
message = ""
while message != 'quit':
    message = input("Tell me something, and I'll repeat it back to you (or type 'quit'): ")
    if message != 'quit':
        print(message)
```

### Using `break` and `continue`


```python
banned_users = ['eve', 'fred', 'gary', 'helen']
players = []

while True:
    player = input("Add a player to your team ('quit' when done): ")
    if player == 'quit':
        break
    elif player in banned_users:
        print(f"{player} is banned!")
        continue
    else:
        players.append(player)

print("Your team:")
for player in players:
    print(player)
```

### Removing all instances of a value from a list


```python
pets = ['dog', 'cat', 'dog', 'fish', 'cat', 'rabbit', 'cat']
print(pets)

while 'cat' in pets:
    pets.remove('cat')

print(pets)
```
