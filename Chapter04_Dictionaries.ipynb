## What are dictionaries?

Python's dictionaries let you connect pieces of related information. Each piece of information
is stored as a **key-value pair**. When you provide a key, Python returns the value associated
with it. Dictionaries use curly braces `{}`, with colons connecting keys and values.


### Making and using a simple dictionary


```python
alien_0 = {'color': 'green', 'points': 5}
print(alien_0['color'])
print(alien_0['points'])
```

### Adding and modifying key-value pairs


```python
alien_0['x_position'] = 0
alien_0['y_position'] = 25
print(alien_0)

alien_0['color'] = 'yellow'
print(alien_0)
```

### Removing key-value pairs


```python
del alien_0['points']
print(alien_0)
```

### Getting a value safely with `.get()`

If a key doesn't exist, `dict['key']` raises a `KeyError`. The `.get()` method returns `None`
(or a default you choose) instead, so it's often safer.


```python
alien_0 = {'color': 'green'}
alien_color = alien_0.get('color')
alien_points = alien_0.get('points', 0)
print(alien_color, alien_points)
```

### Looping through a dictionary


```python
fav_languages = {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'ruby',
    'phil': 'python',
}

print("-- items --")
for name, language in fav_languages.items():
    print(f"{name}: {language}")

print("-- keys --")
for name in fav_languages.keys():
    print(name)

print("-- values --")
for language in fav_languages.values():
    print(language)

print("-- keys, in order --")
for name in sorted(fav_languages.keys()):
    print(name)
```

## Nesting

It's often useful to store a set of dictionaries in a list, or a list inside a dictionary --
this is called **nesting**.


```python
# A list of dictionaries
aliens = [
    {'color': 'green', 'points': 5},
    {'color': 'yellow', 'points': 10},
    {'color': 'red', 'points': 15},
]
for alien in aliens:
    print(alien)

# A dictionary containing a list
fav_languages = {
    'jen': ['python', 'ruby'],
    'sarah': ['c'],
}
for name, langs in fav_languages.items():
    print(f"{name}: {langs}")

# A dictionary containing dictionaries
users = {
    'aeinstein': {'first': 'albert', 'last': 'einstein', 'location': 'princeton'},
    'mcurie': {'first': 'marie', 'last': 'curie', 'location': 'paris'},
}
for username, user_info in users.items():
    full_name = f"{user_info['first']} {user_info['last']}"
    print(f"{username}: {full_name.title()}, {user_info['location'].title()}")
```

## Preserving insertion order

Since Python 3.7, regular dictionaries **remember the order** keys were added in -- you no
longer need `collections.OrderedDict` for that (the old cheat sheet predates this change).
Looping through `fav_languages.items()` above already showed the keys in insertion order.


```python
fav_languages = {}
fav_languages['jen'] = ['python', 'ruby']
fav_languages['sarah'] = ['c']
fav_languages['edward'] = ['ruby', 'go']
fav_languages['phil'] = ['python', 'haskell']

for name, langs in fav_languages.items():
    print(f"{name}: {langs}")
```

### Dictionary comprehensions
Just like list comprehensions, but for dictionaries.


```python
squares = {x: x**2 for x in range(1, 6)}
print(squares)
```
