## Why test your code?

Testing proves that your code works as intended -- both in situations it's designed to handle,
and when people use your programs in unexpected ways. A **unit test** verifies one specific
aspect of your code's behavior. A **test case** is a collection of unit tests.

> **Colab note:** `unittest.main()` normally reads command-line arguments, which don't exist in
> a notebook cell. Call it as `unittest.main(argv=['first-arg-is-ignored'], exit=False)` instead,
> so it runs the tests without trying to exit the whole Python process.


### A function to test


```python
def get_full_name(first, last, middle=''):
    """Return a full name, neatly formatted."""
    if middle:
        full_name = f"{first} {middle} {last}"
    else:
        full_name = f"{first} {last}"
    return full_name.title()

# Quick manual check
print(get_full_name('janis', 'joplin'))
print(get_full_name('david', 'roth', 'lee'))
```

### Testing a function: a passing test


```python
import unittest


class NamesTestCase(unittest.TestCase):
    """Tests for get_full_name()."""

    def test_first_last(self):
        """Names like Janis Joplin work?"""
        full_name = get_full_name('janis', 'joplin')
        self.assertEqual(full_name, 'Janis Joplin')

    def test_middle_name(self):
        """Names like David Lee Roth work?"""
        full_name = get_full_name('david', 'roth', 'lee')
        self.assertEqual(full_name, 'David Lee Roth')


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

### A variety of assert methods

Python's `unittest` module provides several assert methods:

* `assertEqual(a, b)` / `assertNotEqual(a, b)`
* `assertTrue(x)` / `assertFalse(x)`
* `assertIn(item, list)` / `assertNotIn(item, list)`


### Testing a class, with `setUp()`


```python
class Accountant:
    """Manage a bank account."""

    def __init__(self, balance=0):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount


class TestAccountant(unittest.TestCase):
    """Tests for the Accountant class."""

    def setUp(self):
        """Runs before every test method -- gives each test a fresh account."""
        self.acc = Accountant()

    def test_initial_balance(self):
        self.assertEqual(self.acc.balance, 0)

    def test_deposit(self):
        self.acc.deposit(100)
        self.assertEqual(self.acc.balance, 100)
        self.acc.deposit(100)
        self.assertEqual(self.acc.balance, 200)

    def test_withdrawal(self):
        self.acc.deposit(1000)
        self.acc.withdraw(100)
        self.assertEqual(self.acc.balance, 900)


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

### Watching a test fail

Failing tests are important -- they tell you a change broke existing behavior. Here we
introduce a bug (subtracting the withdrawal instead of adding a deposit) and watch the test
catch it.


```python
class BuggyAccountant(Accountant):
    def deposit(self, amount):
        self.balance -= amount   # bug: should be +=


class TestBuggyAccountant(unittest.TestCase):
    def test_deposit(self):
        acc = BuggyAccountant()
        acc.deposit(100)
        self.assertEqual(acc.balance, 100)   # this will fail!


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
