## What are classes?

A class defines the behavior of an object and the kind of information it can store. Data is
held in **attributes**, and functions that belong to a class are called **methods**.


### Creating and using a class


```python
class Car:
    """A simple attempt to model a car."""

    def __init__(self, make, model, year):
        """Initialize car attributes."""
        self.make = make
        self.model = model
        self.year = year
        # Fuel capacity and level in gallons.
        self.fuel_capacity = 15
        self.fuel_level = 0

    def fill_tank(self):
        """Fill gas tank to capacity."""
        self.fuel_level = self.fuel_capacity
        print("Fuel tank is full.")

    def drive(self):
        """Simulate driving."""
        print("The car is moving.")


my_car = Car('audi', 'a4', 2016)
print(my_car.make, my_car.model, my_car.year)

my_car.fill_tank()
my_car.drive()
```

### Modifying attributes directly, or through a method


```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        self.fuel_capacity = 15
        self.fuel_level = 0

    def fill_tank(self):
        """Fill gas tank to capacity."""
        self.fuel_level = self.fuel_capacity
        print("Fuel tank is full.")

    def drive(self):
        """Simulate driving."""
        print("The car is moving.")

    def update_fuel_level(self, new_level):
        """Update the fuel level."""
        if new_level <= self.fuel_capacity:
            self.fuel_level = new_level
        else:
            print("The tank can't hold that much!")

    def add_fuel(self, amount):
        """Add fuel to the tank."""
        if (self.fuel_level + amount) <= self.fuel_capacity:
            self.fuel_level += amount
            print("Added fuel.")
        else:
            print("The tank won't hold that much.")


my_new_car = Car('audi', 'a4', 2016)
my_new_car.fuel_level = 5     # modifying an attribute directly
my_new_car.update_fuel_level(10)
print(my_new_car.fuel_level)
```

## Inheritance

If the class you're writing is a specialized version of another class, you can use
**inheritance**. The child class automatically takes on all attributes and methods of the
parent class, and can override them or add new ones. Use `super().__init__()` to call the
parent's `__init__` method.


```python
class ElectricCar(Car):
    """A simple model of an electric car."""

    def __init__(self, make, model, year):
        """Initialize an electric car."""
        super().__init__(make, model, year)
        # Attributes specific to electric cars.
        self.battery_size = 70
        self.charge_level = 0

    def charge(self):
        """Fully charge the vehicle."""
        self.charge_level = 100
        print("The vehicle is fully charged.")

    def fill_tank(self):
        """Electric cars don't have fuel tanks."""
        print("This car has no fuel tank!")


my_ecar = ElectricCar('tesla', 'model s', 2016)
my_ecar.charge()
my_ecar.drive()
my_ecar.fill_tank()   # overridden method
```

### Instances as attributes

A class can have objects as attributes, which lets classes work together to model more
complex situations.


```python
class Battery:
    """A battery for an electric car."""

    def __init__(self, size=70):
        self.size = size
        self.charge_level = 0

    def get_range(self):
        """Return the battery's range."""
        if self.size == 70:
            return 240
        elif self.size == 85:
            return 270


class ElectricCar(Car):
    def __init__(self, make, model, year):
        super().__init__(make, model, year)
        self.battery = Battery()


my_ecar = ElectricCar('tesla', 'model x', 2016)
print(my_ecar.battery.get_range())
```

### Storing objects in a list

A list can hold as many class instances as you want, which is handy for managing large numbers
of related objects.


```python
gas_fleet = [Car('ford', 'focus', 2016) for _ in range(3)]
electric_fleet = [ElectricCar('nissan', 'leaf', 2016) for _ in range(2)]

for car in gas_fleet:
    car.fill_tank()

for ecar in electric_fleet:
    ecar.battery.charge_level = 100

print("Gas cars:", len(gas_fleet))
print("Electric cars:", len(electric_fleet))
```

### A note on modern Python: dataclasses

For simple classes that mostly just hold data, Python 3.7+ offers the `@dataclass` decorator,
which auto-generates `__init__` and a readable `__repr__` for you. Worth knowing as you go
further, though the explicit `__init__` style above is important to understand first.


```python
from dataclasses import dataclass

@dataclass
class Point:
    x: float
    y: float

p = Point(3, 4)
print(p)          # auto-generated __repr__
print(p.x, p.y)
```
