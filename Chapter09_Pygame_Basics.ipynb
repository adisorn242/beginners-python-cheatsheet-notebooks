## What is Pygame?

Pygame is a framework for making games using Python. It takes care of many lower-level tasks
in building games, letting you focus on the parts of your game that make it interesting.

### Setting up a game window


```python
screen = pg.display.set_mode((1200, 800))
pg.display.set_caption("Alien Invasion")

bg_color = (230, 230, 230)
screen.fill(bg_color)
pg.display.flip()
print("Window created:", screen.get_size())
```

### The game loop

A real game runs an event loop forever, refreshing the screen each pass. Here we run it for a
fixed number of iterations instead of `while True`, so the cell finishes on its own in Colab.
Locally, replace the `for` loop with `while True:` for a real game.


```python
def run_game_demo(frames=5):
    """A stand-in for a real game loop -- runs a few frames, then stops."""
    clock = pg.time.Clock()
    for frame in range(frames):
        for event in pg.event.get():
            if event.type == pg.QUIT:
                return
        screen.fill(bg_color)
        pg.display.flip()
        clock.tick(60)
    print(f"Ran {frames} frames without errors.")

run_game_demo()

# Locally, use this instead for a real, always-on game window:
# while True:
#     for event in pg.event.get():
#         if event.type == pg.QUIT:
#             sys.exit()
#     screen.fill(bg_color)
#     pg.display.flip()
```

## Rect objects

Many game objects are treated as simple rectangles, rather than their actual shape. Pygame's
`Rect` class makes this easy to work with.


```python
screen_rect = screen.get_rect()
print("Screen center:", screen_rect.center)
print("Screen width/height:", screen_rect.width, screen_rect.height)

bullet_rect = pg.Rect(100, 100, 3, 15)
color = (100, 100, 100)
pg.draw.rect(screen, color, bullet_rect)
print("Bullet rect:", bullet_rect)
```

### Responding to keyboard and mouse events

In a real game, these checks live inside the event loop. Shown here as standalone snippets you
can drop into your own loop.


```python
def handle_events():
    for event in pg.event.get():
        if event.type == pg.KEYDOWN:
            if event.key == pg.K_RIGHT:
                print("move right")
            elif event.key == pg.K_LEFT:
                print("move left")
            elif event.key == pg.K_SPACE:
                print("fire!")
            elif event.key == pg.K_q:
                print("quit requested")
        elif event.type == pg.KEYUP:
            if event.key == pg.K_RIGHT:
                print("stop moving right")
        elif event.type == pg.MOUSEBUTTONDOWN:
            mouse_pos = pg.mouse.get_pos()
            print("clicked at", mouse_pos)

print("handle_events() defined -- call it inside your real game loop.")
```

### Groups and collision detection

A `Group` is like a list with extra game-related functionality. It's the standard way to manage
many similar objects (bullets, aliens, etc.) at once.


```python
from pygame.sprite import Sprite, Group


class Bullet(Sprite):
    """A simple bullet fired from a ship."""

    def __init__(self, x, y):
        super().__init__()
        self.image = pg.Surface((3, 15))
        self.image.fill((60, 60, 60))
        self.rect = self.image.get_rect(center=(x, y))

    def update(self):
        self.rect.y -= 1


bullets = Group()
bullets.add(Bullet(600, 400))
bullets.add(Bullet(610, 400))

bullets.update()
print("Number of bullets:", len(bullets))

# Detecting collisions between a group and a single sprite:
# if pg.sprite.spritecollideany(ship, aliens):
#     ships_left -= 1

# Detecting collisions between two groups:
# collisions = pg.sprite.groupcollide(bullets, aliens, True, True)
```
