## What are variables?

A variable is a name you give to a piece of data so you can use it later. Python figures out
the *type* of data automatically -- you never have to declare it yourself. A **string** is a
series of characters, surrounded by single or double quotes.


### Hello, world!
Every language tradition starts here.


```python
print("Hello world!")
```

### Storing a message in a variable
A variable acts like a label pointing at a value.


```python
msg = "Hello world!"
print(msg)
```

### Concatenation (combining strings)

You can join strings together with `+`. In modern Python it's usually nicer to use an
**f-string** instead (see below), but plain concatenation still works and is worth knowing.


```python
first_name = 'albert'
last_name = 'einstein'
full_name = first_name + ' ' + last_name
print(full_name)
```

### f-strings (the modern way to build strings)

The original cheat sheet uses `+` concatenation and the old `.format()` method. Since Python
3.6, **f-strings** are the recommended way to build strings -- they're shorter and easier to
read. We'll use them throughout this notebook series.


```python
first_name = 'albert'
last_name = 'einstein'
full_name = f"{first_name} {last_name}"
print(full_name)
print(f"Hello, {full_name.title()}!")
```

### Useful string methods

A few methods you'll use constantly: `.title()`, `.upper()`, `.lower()`, and `.strip()`.


```python
name = "  ada lovelace  "
print(name.title())
print(name.upper())
print(name.lower())
print(name.strip())
```

## Numbers

Python has integers (`int`) and floating-point numbers (`float`). Basic arithmetic works as
you'd expect, and you can mix ints and floats freely -- Python converts automatically.


```python
print(2 + 3)      # addition
print(3 - 2)      # subtraction
print(2 * 3)      # multiplication
print(3 / 2)      # division -> float
print(3 // 2)     # floor division -> int
print(3 ** 2)     # exponent
print(3 % 2)      # modulo (remainder)
```

### Underscores in large numbers

Since Python 3.6, you can use underscores to group digits in large numbers for readability.
Python ignores the underscores.


```python
universe_age = 14_000_000_000
print(universe_age)
```
