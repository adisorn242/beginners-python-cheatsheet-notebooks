## What is Pygal?

Pygal renders charts as SVG files, which look great and support interactive elements when
viewed in a browser. In Colab we'll render each chart to a file, then display it inline with
`IPython.display.SVG` -- no `force_uri_protocol` setting needed anymore for that.


```python
!pip install -q pygal
```

```python
import pygal
from IPython.display import SVG, display

print("Pygal version:", pygal.__version__)
```

### Making a line graph


```python
x_values = [0, 1, 2, 3, 4, 5]
squares = [0, 1, 4, 9, 16, 25]

chart = pygal.Line()
chart.title = "Squares"
chart.x_labels = x_values
chart.add('Squares', squares)
chart.render_to_file('squares.svg')

display(SVG('squares.svg'))
```

### Making a scatter plot


```python
squares = [(x, x**2) for x in range(11)]

chart = pygal.XY(stroke=False)
chart.title = "Squares"
chart.add('x^2', squares)
chart.render_to_file('squares_scatter.svg')

display(SVG('squares_scatter.svg'))
```

### Making a bar graph, from a list or a dictionary


```python
outcomes = [1, 2, 3, 4, 5, 6]
frequencies = [18, 16, 18, 17, 18, 13]

chart = pygal.Bar()
chart.title = "Results of rolling one D6 1,000 times"
chart.x_labels = outcomes
chart.add('D6', frequencies)
chart.render_to_file('rolling_dice.svg')

display(SVG('rolling_dice.svg'))

# The same chart, built from a dictionary instead:
results = {1: 18, 2: 16, 3: 18, 4: 17, 5: 18, 6: 13}
chart = pygal.Bar()
chart.x_labels = results.keys()
chart.add('D6', results.values())
chart.render_to_file('rolling_dice_dict.svg')
display(SVG('rolling_dice_dict.svg'))
```

### Plotting multiple data series, and filling the area under a series


```python
x_values = list(range(11))
squares = [x**2 for x in x_values]
cubes = [x**3 for x in x_values]

chart = pygal.Line(fill=True)
chart.title = "Squares and Cubes"
chart.x_labels = x_values
chart.add('Squares', squares)
chart.add('Cubes', cubes)
chart.render_to_file('squares_cubes.svg')

display(SVG('squares_cubes.svg'))
```

### Styling plots with built-in and custom styles


```python
from pygal.style import LightGreenStyle, LightenStyle

chart_style = LightGreenStyle()
chart = pygal.Line(style=chart_style)
chart.title = "Squares and Cubes"
chart.x_labels = x_values
chart.add('Squares', squares)
chart.add('Cubes', cubes)
chart.render_to_file('styled_squares_cubes.svg')
display(SVG('styled_squares_cubes.svg'))

chart_style = LightenStyle('#336688')
chart = pygal.Bar(style=chart_style)
chart.x_labels = ['django', 'requests', 'scikit-learn']
chart.y_title = 'Stars'
chart.add('Python Repos', [20506, 20054, 12607])
chart.render_to_file('python_repos.svg')
display(SVG('python_repos.svg'))
```

## World maps (optional -- may not install)

Pygal's world-map support lives in a separate, community-maintained package,
`pygal_maps_world`, which can be slow to keep up with new Pygal releases. This cell is wrapped
in a `try`/`except` so the rest of the notebook keeps working even if the install fails.


```python
try:
    get_ipython().system('pip install -q pygal_maps_world')
    from pygal.maps.world import World

    wm = World()
    wm.title = 'North America'
    wm.add('North America', ['ca', 'mx', 'us'])
    wm.render_to_file('north_america.svg')
    display(SVG('north_america.svg'))
except Exception as e:
    print("Skipping world map -- pygal_maps_world isn't available:", e)
```
