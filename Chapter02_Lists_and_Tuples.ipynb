## What are lists?

A list stores a series of items in a particular order. Lists let you store sets of information
in one place, whether you have a few items or millions. Use square brackets `[]` to define a
list, and commas to separate individual items.


### Making a list and accessing elements


```python
bikes = ['trek', 'redline', 'giant']
print(bikes)

first_bike = bikes[0]
last_bike = bikes[-1]
print(first_bike, last_bike)
```

### Looping through a list


```python
for bike in bikes:
    print(bike)
```

### Adding, inserting, and removing items


```python
bikes = []
bikes.append('trek')
bikes.append('redline')
bikes.append('giant')
bikes.insert(0, 'schwinn')
print(bikes)

del bikes[-1]
print(bikes)

removed = bikes.pop()   # removes & returns the last item
print("removed:", removed, "-> now:", bikes)

bikes.remove('schwinn')  # removes the first matching value
print(bikes)
```

### Making numerical lists with range()


```python
numbers = list(range(1, 11))
print(numbers)

squares = [x**2 for x in range(1, 11)]
print(squares)
```

### List comprehensions

A comprehension builds a list in one line: define an expression for the values you want, then
a `for` loop to generate the inputs.


```python
squares = [x**2 for x in range(1, 11)]
print(squares)

names = ['kai', 'abe', 'ada', 'gus', 'zoe']
upper_names = [name.upper() for name in names]
print(upper_names)
```

### Slicing a list


```python
finishers = ['sam', 'bob', 'ada', 'bea']
first_two = finishers[:2]
middle_two = finishers[1:3]
last_two = finishers[-2:]
print(first_two, middle_two, last_two)
```

### Copying a list

Slicing with `[:]` makes a real copy. If you just assign `new_list = old_list`, both names
point at the *same* list -- changing one changes the other!


```python
bikes = ['trek', 'redline', 'giant']
copy_of_bikes = bikes[:]
copy_of_bikes.append('surly')

print(bikes)
print(copy_of_bikes)
```

### Sorting a list


```python
users = ['val', 'bob', 'mia', 'ron', 'ned']

print(sorted(users))                 # temporary sort, doesn't change users
print(sorted(users, reverse=True))
print(users)                         # unchanged

users.sort()                         # permanent sort
print(users)

users.reverse()                      # reverses order in place
print(users)
```

### Simple statistics on a list of numbers


```python
ages = [93, 99, 66, 17, 85, 1, 35, 82, 2, 77]
print(min(ages), max(ages), sum(ages))
```

## Tuples

A tuple is like a list, but its items **can't be changed** once defined (it's *immutable*).
Tuples use parentheses `()` instead of square brackets, and are good for values that shouldn't
change, like screen dimensions.


```python
dimensions = (1920, 1080)
for dimension in dimensions:
    print(dimension)

# You can overwrite the whole tuple, just not individual items
dimensions = (1200, 900)
print(dimensions)

try:
    dimensions[0] = 400
except TypeError as e:
    print("Error:", e)
```
