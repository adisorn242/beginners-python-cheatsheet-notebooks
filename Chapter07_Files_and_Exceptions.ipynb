## Reading from a file

To read from a file, open it and read its contents. The `with` statement makes sure the file
is closed properly once you're done with it.

> **Colab note:** files you create in a Colab notebook live on a temporary virtual machine and
> disappear when the runtime resets. That's fine for this notebook -- we'll create a small
> sample file first so every example below is self-contained and runnable.


```python
# Create a sample file to read from, so this notebook is self-contained.
sample_text = "The quiet interstellar dust.\nThe silent depths of space.\nThe endless night sky.\n"
with open('sample.txt', 'w') as f:
    f.write(sample_text)

print("sample.txt created.")
```

### Reading an entire file at once


```python
filename = 'sample.txt'

with open(filename) as f_obj:
    contents = f_obj.read()

print(contents)
```

### Reading line by line, and storing lines in a list


```python
with open(filename) as f_obj:
    for line in f_obj:
        print(line.rstrip())   # rstrip() removes the trailing newline

with open(filename) as f_obj:
    lines = f_obj.readlines()

for line in lines:
    print(line.rstrip())
```

### Writing to a file

Passing `'w'` to `open()` tells Python you want to write. Be careful -- this erases the file's
existing contents if it already exists! Use `'a'` to append instead.


```python
filename = 'programming.txt'

with open(filename, 'w') as f:
    f.write("I love programming.\n")
    f.write("I love creating new games.\n")

with open(filename, 'a') as f:
    f.write("I also love working with data.\n")
    f.write("I love making apps as well.\n")

with open(filename) as f:
    print(f.read())
```

### File paths

`open()` looks for the file in the current working directory by default. You can also give a
relative or absolute path. On Windows, use backslashes (or a raw string) in paths.


```python
import os

os.makedirs('text_files', exist_ok=True)
with open('text_files/nested_example.txt', 'w') as f:
    f.write("This file lives in a subfolder.\n")

with open('text_files/nested_example.txt') as f:
    print(f.read())

# Example Windows-style path (not run, just illustrative):
# f_path = r"C:\Users\ehmatthes\books\alice.txt" 
```

## Exceptions

Exceptions are special objects Python creates to handle errors gracefully. You place code that
might cause an error inside a `try` block; the `except` block runs if that error occurs; an
optional `else` block runs only if the `try` block succeeded.


```python
print("Enter two numbers, and I'll divide them.")

try:
    x = int(input("First number: "))
    y = int(input("Second number: "))
    result = x / y
except ZeroDivisionError:
    print("You can't divide by zero!")
except ValueError:
    print("Please enter valid whole numbers.")
else:
    print(result)
```

### Handling a missing file with `FileNotFoundError`

Avoid **bare** `except:` blocks -- they catch *everything*, including errors you didn't
anticipate. Catch specific exceptions instead.


```python
f_name = 'does_not_exist.txt'

try:
    with open(f_name) as f_obj:
        lines = f_obj.readlines()
except FileNotFoundError:
    print(f"Sorry, the file {f_name} does not exist.")
else:
    print(lines)
```

### Failing silently with `pass`

Sometimes you want a program to continue running without reporting every error to the user.


```python
f_names = ['sample.txt', 'programming.txt', 'moby_dick.txt', 'little_women.txt']

for f_name in f_names:
    try:
        with open(f_name) as f_obj:
            lines = f_obj.readlines()
    except FileNotFoundError:
        pass   # just move on to the next file
    else:
        print(f"{f_name} has {len(lines)} lines.")
```

## Storing data with JSON

The `json` module lets you dump simple Python data structures into a file, and load them back
later. JSON isn't specific to Python, so it's a great way to share data with other languages.


```python
import json

numbers = [2, 3, 5, 7, 11, 13]

filename = 'numbers.json'
with open(filename, 'w') as f_obj:
    json.dump(numbers, f_obj)

with open(filename) as f_obj:
    loaded_numbers = json.load(f_obj)

print(loaded_numbers)
```
